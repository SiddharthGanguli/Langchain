```python
from langchain_community.llms import Ollama
from langchain_core.messages import HumanMessage,SystemMessage
from langchain.chains import ConversationChain
from langchain_core.prompts import ChatPromptTemplate
from langchain.memory import ConversationBufferMemory
```

```python
model=Ollama(model='gemma:2b')
```

```python
template= ChatPromptTemplate.from_messages([
    ('system','you are a helpfull ai agent , your name is Landu.'),
    ('human','content {content}'),
    ('human','question {question}')
])
```

```python
prompt = template.invoke({
    'content': '''I am a software engineer with 5 years of experience in developing web applications.
I have worked with a variety of technologies including Python, JavaScript, and SQL.
I am looking to transition into a DevOps role and am interested in learning more about the field.''',
    'question': 'What are the key responsibilities of a DevOps engineer?'
})
```

```python
memory=ConversationBufferMemory()
```

```python
conversation = ConversationChain(
    llm=model,
    memory=memory,
    verbose=True
)
```

```python
print(conversation.invoke(prompt))
print(conversation.invoke("What is my experience in software engineering "))
```

```python
from langchain_community.llms import Ollama
from langchain.chains import LLMChain
from langchain_core.prompts import ChatPromptTemplate
from langchain.memory.buffer import ConversationBufferMemory

# Load model
llm = Ollama(model='gemma:2b')

# Setup prompt
template = ChatPromptTemplate.from_messages([
    ('system', 'You are a helpful AI named Landu.'),
    ('human', '{input}')
])

# Setup memory
memory = ConversationBufferMemory(return_messages=True)

# Build chain
chain = LLMChain(
    llm=llm,
    prompt=template,
    memory=memory,
    verbose=True
)

# Interact
chain.invoke({"input": "I am a software engineer with 5 years of experience in web development."})
response = chain.invoke({"input": "What is my experience in software engineering?"})
print(response["text"])
```
