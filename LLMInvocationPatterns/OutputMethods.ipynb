```python
from  langchain_community.llms import Ollama
from langchain_core.messages import HumanMessage ,AIMessage ,SystemMessage
```

```python
model=Ollama(model='llama3:latest')
```

### Invoke : Simple Q&A or one prompt at a time

```python
output=model.invoke("What is the capital of india? ")
print(f"Ai {output}")
```

### batch: 	Fast processing of multiple prompts

```python
prompts = [
    "What is AI?",
    "What is Machine Learning?",
    "What is Deep Learning?"
]
```

```python
model=Ollama(model='gemma:2b')
responses = model.batch(prompts)


for i in responses:
    print(i,'\n')
```

### stream:	Chatbot UI or anything real-time

```python
llm=Ollama(model='gemma:2b')
for chunk in llm.stream("Tell me a story about virat kohli."):
    print(chunk, end="", flush=True)
```
