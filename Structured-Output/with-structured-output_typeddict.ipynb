```python
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
from typing import TypedDict

```

```python
load_dotenv()
```

```python
model = ChatOpenAI(model='gpt-3.5-turbo')
```

defining the schema

```python
class reviw(TypedDict):
    key_terms:str
    summary:str
    sentiment:str
    pros:str
    cons:str
```

```python
structured_output=model.with_structured_output(reviw)
```

```python
result=structured_output.invoke('''I recently upgraded to the Samsung Galaxy S24 Ultra, and I must say, it’s an absolute powerhouse! The Snapdragon 8 Gen 3 processor makes everything lightning fast—whether I’m gaming, multitasking, or editing photos. The 5000mAh battery easily lasts a full day even with heavy use, and the 45W fast charging is a lifesaver.

The S-Pen integration is a great touch for note-taking and quick sketches, though I don't use it often. What really blew me away is the 200MP camera—the night mode is stunning, capturing crisp, vibrant images even in low light. Zooming up to 100x actually works well for distant objects, but anything beyond 30x loses quality.

However, the weight and size make it a bit uncomfortable for one-handed use. Also, Samsung’s One UI still comes with bloatware—why do I need five different Samsung apps for things Google already provides? The $1,300 price tag is also a hard pill to swallow.

Pros:
Insanely powerful processor (great for gaming and productivity)
Stunning 200MP camera with incredible zoom capabilities
Long battery life with fast charging
S-Pen support is unique and useful''')
```

```python
print(result)
```

in typing there is a function called Annotated<br>
this are used to give a description for that specific key 

```python
from typing import Annotated,List


class review(TypedDict):
    key_terms:Annotated[str,"Write down all the key themes discussed in the review in a list"]
    summary:Annotated[str,"A brief summary of the review"]
    sentiment: Annotated[str,"Return sentiment of the review either negative, positive or neutral"]
    pros: Annotated[str,"Write down all the pros."]
    cros: Annotated[str,"Write down all the cons."]

structured_model = model.with_structured_output(review)
```

```python
result=structured_model.invoke("I love the keyboard on this laptop. It's clicky and responsive. The battery life is solid too. But the screen is a little dim, and the speakers are just okay.")

```

```python
from typing import Annotated,Optional


class review(TypedDict):
    key_terms:Annotated[str,"Write down all the key themes discussed in the review in a list"]
    summary:Annotated[str,"A brief summary of the review"]
    sentiment: Annotated[str,"Return sentiment of the review either negative, positive or neutral"]
    pros: Annotated[Optional[str],"Write down all the pros."]
    cros: Annotated[Optional[str],"Write down all the cons."]
    name: Annotated[Optional[str],"Write the name of the reviewer"]

structured_model = model.with_structured_output(review)


result=structured_model.invoke( "I love the keyboard on this laptop. It's clicky and responsive. The battery life is solid too. But the screen is a little dim, and the speakers are just okay.")

```
